# Email Hunt

Emails are everywhere, when you sign up for a website, apply for a job, or send a message. But can you detect emails in messy text with regex?

Here's a block of nonsense text with some emails hidden inside

```python
s = """Thanks for reaching out! You can email us at info@example.com.  
For support, try support@my-website.org or helpdesk@my-website.org.  
Our sales team uses sales@shop123.co.uk, while marketing prefers marketing.team@company.net.  

Some people mistype, like hello@example, user@domain, test@123..com, or noatsign.com.  
Watch out—your regex should only catch real emails!  

Other valid addresses include ceo@bigcorp.com, hello.world@startup.io,  
and first.last@university.edu.  

Fake trick: admin@localhost or name@company, not real!  
But these should work: contact@nonprofit.org and newsletter@updates.co.  
"""
```

## Step 1: Simple Match

Write a regex expression that matches the simplest case of an email

```python
import re

def reg(regex, text):
    print(re.sub(f'({regex})', r'\033[1;30;43m\1\033[m', text))
    
reg(r"something_here", s)
```

**Hint:** Try something like `\w+@\w+\.\w+` first. What happens? Does it catch everything? Does it catch _too_ much?

## Step 3: Count Unique Domains

Use `re.findall` to extract all emails. Then count how many emails belong to each domain (like `example.com`, `my-website.org`, etc.):

```python
import pandas as pd

emails = re.findall(r"????", s) # replace expression here
domains = {}

for e in emails:
    dom = e.split("@")[1]
    domains[dom] = domains.get(dom, 0) + 1

pd.Series(domains).sort_values().plot.barh(fontsize=20, color="k")
```
